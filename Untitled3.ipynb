```python
from google.colab import userdata
from groq import Groq

client = Groq(api_key=userdata.get('GROQ_API_KEY'))
```

```python
import gradio as gr

def generate_quiz(text, question_type, num_questions):
    if not text.strip():
        return "⚠️ Please enter some text first."
    try:
        prompt = f"""
You are a professional quiz generator.
Based on the following text, generate {num_questions} {question_type} questions.

Format the output clearly:
- Number each question
- For MCQ: provide 4 options (A, B, C, D) and mark the correct answer
- For True/False: state the answer clearly
- For Short Answer: provide a model answer

Text:
{text}

Generate the quiz now:
"""
        response = client.chat.completions.create(
            model="llama-3.1-8b-instant",
            messages=[{"role": "user", "content": prompt}],
            max_tokens=1000
        )
        return response.choices[0].message.content
    except Exception as e:
        return f"Error: {str(e)}"

with gr.Blocks(title="AI Quiz Generator", theme=gr.themes.Soft()) as app:
    gr.Markdown("# 🎓 AI Quiz Generator")
    gr.Markdown("Paste any text and generate quiz questions instantly using AI.")

    with gr.Row():
        with gr.Column():
            text_input = gr.Textbox(
                label="📄 Input Text",
                placeholder="Paste your study material here...",
                lines=10
            )
            with gr.Row():
                q_type = gr.Dropdown(
                    choices=["Multiple Choice (MCQ)", "True/False", "Short Answer"],
                    value="Multiple Choice (MCQ)",
                    label="Question Type"
                )
                q_num = gr.Slider(
                    minimum=1, maximum=10, value=5, step=1,
                    label="Number of Questions"
                )
            submit_btn = gr.Button("🚀 Generate Quiz", variant="primary")

        with gr.Column():
            output = gr.Textbox(
                label="📝 Generated Quiz",
                lines=20,
                show_copy_button=True
            )

    submit_btn.click(
        fn=generate_quiz,
        inputs=[text_input, q_type, q_num],
        outputs=output
    )

app.launch(share=True)
```
