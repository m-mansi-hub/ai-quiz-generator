```python
import gradio as gr
import json, re

custom_css = """
* { font-family: 'Segoe UI', sans-serif; }
body, .gradio-container { background: #f8fafc !important; }

#header {
    background: #0f4c75; border-radius: 14px;
    padding: 22px 28px; margin-bottom: 16px;
    display: flex; align-items: center; gap: 16px;
}
#header h1 { color: #fff !important; font-size: 1.4rem !important; font-weight: 700 !important; margin: 0 0 4px !important; }
#header p  { color: #a8d4f0 !important; font-size: 0.85rem !important; margin: 0 !important; }

.panel { background: #fff !important; border: 1px solid #e2e8f0 !important; border-radius: 14px !important; padding: 20px !important; }

label { font-size: 0.72rem !important; font-weight: 700 !important; letter-spacing: 0.07em !important; text-transform: uppercase !important; color: #64748b !important; }

textarea { background: #f8fafc !important; border: 1.5px solid #e2e8f0 !important; border-radius: 10px !important; color: #1e293b !important; font-size: 0.92rem !important; line-height: 1.7 !important; }
textarea:focus { border-color: #1abc9c !important; box-shadow: 0 0 0 3px rgba(26,188,156,0.12) !important; }

#gen-btn, #sum-btn, #submit-btn, #next-btn {
    background: #0f4c75 !important;
    border: none !important;
    border-radius: 10px !important;
    color: #fff !important;
    font-size: 0.95rem !important;
    font-weight: 700 !important;
    padding: 13px !important;
    width: 100% !important;
}

#gen-btn:hover, #sum-btn:hover, #submit-btn:hover, #next-btn:hover {
    opacity: 0.88 !important;
}

.footer-text {
    text-align: center;
    color: #94a3b8;
    font-size: 0.76rem;
    margin-top: 14px;
}
"""

# ── أدوات مساعدة ─────────────────────────────────────────────

def is_arabic(text):
    count = sum(1 for c in text if '\u0600' <= c <= '\u06FF')
    return count > len(text) * 0.25


def extract_json_array(raw):
    start = raw.find("[")
    if start == -1:
        return None

    depth = 0
    for i in range(start, len(raw)):
        if raw[i] == "[":
            depth += 1
        elif raw[i] == "]":
            depth -= 1
            if depth == 0:
                return raw[start:i+1]

    return None


def validate_questions(questions):
    clean_questions = []

    if not isinstance(questions, list):
        return clean_questions

    for q in questions:
        if not isinstance(q, dict):
            continue

        if not all(k in q for k in ["question", "options", "correct"]):
            continue

        if not isinstance(q["options"], list) or len(q["options"]) < 2:
            continue

        try:
            q["correct"] = int(q["correct"])
        except:
            continue

        if q["correct"] < 0 or q["correct"] >= len(q["options"]):
            continue

        clean_questions.append(q)

    return clean_questions


# ── توليد الأسئلة ─────────────────────────────────────────────

def generate_quiz(text, question_type, num_questions):
    if not text.strip():
        return (
            [],
            0,
            0,
            "⚠️ الرجاء إدخال نص أولاً.",
            "السؤال سيظهر هنا.",
            gr.update(choices=[], value=None),
            ""
        )

    arabic = is_arabic(text)
    lang_note = "اكتب الأسئلة والخيارات باللغة العربية." if arabic else "Write questions and options in English."

    if question_type == "True/False":
        if arabic:
            fmt = '[{"question":"...","options":["صحيح","خطأ"],"correct":0}]'
        else:
            fmt = '[{"question":"...","options":["True","False"],"correct":0}]'
    else:
        fmt = '[{"question":"...","options":["A","B","C","D"],"correct":0}]'

    prompt = f"""أنت مولّد اختبارات محترف. {lang_note}
بناءً على النص التالي، أنشئ {int(num_questions)} سؤال من نوع {question_type}.

أعد JSON Array فقط.
لا تكتب أي شرح.
لا تكتب Markdown.
لا تكرر JSON أكثر من مرة.

الشكل المطلوب:
{fmt}

مهم:
- correct يجب أن يكون رقم الخيار الصحيح ويبدأ من 0.
- لا تكتب أي شيء قبل أو بعد JSON.
- الأسئلة يجب أن تكون واضحة ومباشرة.
- الخيارات يجب أن تكون منطقية.

النص:
{text}"""

    try:
        response = client.chat.completions.create(
            model="llama-3.1-8b-instant",
            messages=[
                {
                    "role": "system",
                    "content": "You only return valid JSON arrays. No explanations. No markdown."
                },
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            max_tokens=2000,
            temperature=0.2
        )

        raw = response.choices[0].message.content
        json_text = extract_json_array(raw)

        if not json_text:
            return (
                [],
                0,
                0,
                "❌ خطأ في تحليل الإجابة، حاول مرة أخرى.",
                "السؤال سيظهر هنا.",
                gr.update(choices=[], value=None),
                ""
            )

        questions = json.loads(json_text)
        questions = validate_questions(questions)

        if not questions:
            return (
                [],
                0,
                0,
                "❌ لم يتم توليد أسئلة صالحة، حاول مرة أخرى.",
                "السؤال سيظهر هنا.",
                gr.update(choices=[], value=None),
                ""
            )

        first_q = questions[0]
        question_text = f"Q1. {first_q['question']}"

        return (
            questions,
            0,
            0,
            f"✅ تم توليد {len(questions)} سؤال بنجاح.",
            question_text,
            gr.update(choices=first_q["options"], value=None),
            ""
        )

    except Exception as e:
        return (
            [],
            0,
            0,
            f"❌ خطأ: {str(e)}",
            "السؤال سيظهر هنا.",
            gr.update(choices=[], value=None),
            ""
        )


# ── فحص الإجابة ─────────────────────────────────────────────

def submit_answer(answer, questions, index, score):
    if not questions:
        return "⚠️ لا يوجد أسئلة بعد. اضغط Generate Quiz أولاً.", score

    if answer is None:
        return "⚠️ اختر إجابة أولاً.", score

    current_q = questions[index]
    correct_index = current_q["correct"]
    correct_answer = current_q["options"][correct_index]

    if answer == correct_answer:
        score += 1
        return "✅ إجابة صحيحة!", score
    else:
        return f"❌ إجابة خاطئة. الإجابة الصحيحة هي: {correct_answer}", score


# ── السؤال التالي ─────────────────────────────────────────────

def next_question(questions, index, score):
    if not questions:
        return (
            index,
            "السؤال سيظهر هنا.",
            gr.update(choices=[], value=None),
            ""
        )

    index += 1

    if index >= len(questions):
        final_text = f"🎉 انتهى الاختبار. نتيجتك: {score} / {len(questions)}"
        return (
            index,
            final_text,
            gr.update(choices=[], value=None),
            ""
        )

    q = questions[index]
    question_text = f"Q{index + 1}. {q['question']}"

    return (
        index,
        question_text,
        gr.update(choices=q["options"], value=None),
        ""
    )


# ── تلخيص النص ─────────────────────────────────────────────

def summarize_text(text):
    if not text.strip():
        return "⚠️ الرجاء إدخال نص أولاً."

    arabic = is_arabic(text)

    direction_note = "أجب باللغة العربية واستخدم النقاط المرقّمة." if arabic else "Answer in English using bullet points."

    prompt = f"""أنت مساعد ذكي متخصص في تلخيص النصوص. {direction_note}
قدّم:
1. ملخص موجز من 3 إلى 4 جمل
2. أهم النقاط على شكل قائمة
3. الكلمات المفتاحية

النص:
{text}"""

    try:
        response = client.chat.completions.create(
            model="llama-3.1-8b-instant",
            messages=[
                {
                    "role": "system",
                    "content": "You are a helpful text summarizer."
                },
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            max_tokens=1000,
            temperature=0.3
        )

        return response.choices[0].message.content

    except Exception as e:
        return f"❌ خطأ: {str(e)}"


# ── الواجهة ─────────────────────────────────────────────

with gr.Blocks(css=custom_css, title="AI Quiz Generator") as app:

    questions_state = gr.State([])
    index_state = gr.State(0)
    score_state = gr.State(0)

    gr.HTML("""
    <div id="header">
        <div style="width:48px;height:48px;background:#1abc9c;border-radius:12px;
                    display:flex;align-items:center;justify-content:center;
                    font-size:22px;font-weight:700;color:#fff;flex-shrink:0">Q</div>
        <div>
            <h1>AI Quiz Generator</h1>
            <p>Generate quizzes and summaries — supports Arabic & English</p>
        </div>
    </div>
    """)

    with gr.Tabs():

        with gr.TabItem("🎯 Quiz Generator"):
            with gr.Row():
                with gr.Column(elem_classes="panel"):
                    text_input = gr.Textbox(
                        label="📄 Input Text — يدعم العربية والإنجليزية",
                        placeholder="الصق النص هنا... أو Paste your study material here...",
                        lines=10
                    )

                    with gr.Row():
                        q_type = gr.Dropdown(
                            choices=["Multiple Choice (MCQ)", "True/False"],
                            value="Multiple Choice (MCQ)",
                            label="Question Type"
                        )

                        q_num = gr.Slider(
                            minimum=1,
                            maximum=10,
                            value=5,
                            step=1,
                            label="Number of Questions"
                        )

                    gen_btn = gr.Button("🚀 Generate Quiz", elem_id="gen-btn")

                with gr.Column(elem_classes="panel"):
                    status = gr.Markdown("اضغط Generate لتوليد الأسئلة.")
                    question_box = gr.Markdown("السؤال سيظهر هنا.")
                    answer_radio = gr.Radio(
                        choices=[],
                        label="Choose your answer",
                        interactive=True
                    )

                    with gr.Row():
                        submit_btn = gr.Button("✅ Submit Answer", elem_id="submit-btn")
                        next_btn = gr.Button("➡️ Next Question", elem_id="next-btn")

                    feedback = gr.Markdown("")

        with gr.TabItem("📄 Text Summarizer"):
            with gr.Row():
                with gr.Column(elem_classes="panel"):
                    sum_input = gr.Textbox(
                        label="📄 النص للتلخيص — يدعم العربية والإنجليزية",
                        placeholder="الصق التقرير أو الفقرة هنا...",
                        lines=12
                    )

                    sum_btn = gr.Button("📋 Summarize", elem_id="sum-btn")

                with gr.Column(elem_classes="panel"):
                    sum_output = gr.Textbox(
                        label="📝 Summary",
                        lines=15,
                        show_copy_button=True
                    )

    gr.HTML('<div class="footer-text">COMP 491 · Almaarefa University · Powered by LLaMA 3.1 via Groq</div>')

    gen_btn.click(
        fn=generate_quiz,
        inputs=[text_input, q_type, q_num],
        outputs=[
            questions_state,
            index_state,
            score_state,
            status,
            question_box,
            answer_radio,
            feedback
        ]
    )

    submit_btn.click(
        fn=submit_answer,
        inputs=[answer_radio, questions_state, index_state, score_state],
        outputs=[feedback, score_state]
    )

    next_btn.click(
        fn=next_question,
        inputs=[questions_state, index_state, score_state],
        outputs=[index_state, question_box, answer_radio, feedback]
    )

    sum_btn.click(
        fn=summarize_text,
        inputs=[sum_input],
        outputs=[sum_output]
    )

app.launch(share=True)
```

```python
import gradio as gr
import json, re

custom_css = """
@import url('https://fonts.googleapis.com/css2?family=Inter:wght@400;500;600;700;800&display=swap');

* {
    font-family: 'Inter', 'Segoe UI', sans-serif !important;
}

body, .gradio-container {
    background:
        radial-gradient(circle at top left, rgba(56, 189, 248, 0.18), transparent 32%),
        radial-gradient(circle at top right, rgba(139, 92, 246, 0.18), transparent 28%),
        linear-gradient(135deg, #020617 0%, #0f172a 45%, #111827 100%) !important;
    color: #e5e7eb !important;
}

.gradio-container {
    max-width: 1180px !important;
    margin: auto !important;
    padding: 22px !important;
}

#hero {
    background:
        linear-gradient(135deg, rgba(15, 23, 42, 0.92), rgba(30, 41, 59, 0.88)),
        linear-gradient(135deg, #38bdf8, #8b5cf6);
    border: 1px solid rgba(148, 163, 184, 0.22);
    border-radius: 26px;
    padding: 30px;
    margin-bottom: 22px;
    box-shadow: 0 24px 70px rgba(0, 0, 0, 0.35);
    position: relative;
    overflow: hidden;
}

#hero::before {
    content: "";
    position: absolute;
    width: 220px;
    height: 220px;
    background: rgba(56, 189, 248, 0.22);
    border-radius: 50%;
    right: -70px;
    top: -90px;
    filter: blur(8px);
}

#hero-content {
    position: relative;
    z-index: 2;
    display: flex;
    align-items: center;
    gap: 18px;
}

.logo-box {
    width: 64px;
    height: 64px;
    border-radius: 20px;
    background: linear-gradient(135deg, #38bdf8, #8b5cf6);
    color: white;
    display: flex;
    align-items: center;
    justify-content: center;
    font-size: 28px;
    font-weight: 800;
    box-shadow: 0 14px 35px rgba(56, 189, 248, 0.35);
}

.hero-title {
    color: #ffffff !important;
    font-size: 2rem !important;
    font-weight: 800 !important;
    margin: 0 !important;
    letter-spacing: -0.04em;
}

.hero-subtitle {
    color: #cbd5e1 !important;
    font-size: 0.98rem !important;
    margin-top: 7px !important;
    line-height: 1.7;
}

.badges {
    display: flex;
    flex-wrap: wrap;
    gap: 8px;
    margin-top: 14px;
}

.badge {
    background: rgba(255,255,255,0.08);
    border: 1px solid rgba(255,255,255,0.16);
    color: #e0f2fe;
    padding: 6px 11px;
    border-radius: 999px;
    font-size: 0.74rem;
    font-weight: 600;
}

.panel {
    background: rgba(15, 23, 42, 0.72) !important;
    border: 1px solid rgba(148, 163, 184, 0.20) !important;
    border-radius: 24px !important;
    padding: 22px !important;
    box-shadow: 0 20px 60px rgba(0, 0, 0, 0.28) !important;
    backdrop-filter: blur(18px);
}

label {
    color: #93c5fd !important;
    font-size: 0.74rem !important;
    font-weight: 800 !important;
    letter-spacing: 0.08em !important;
    text-transform: uppercase !important;
}

textarea, input, select {
    background: rgba(2, 6, 23, 0.72) !important;
    border: 1.5px solid rgba(148, 163, 184, 0.28) !important;
    border-radius: 16px !important;
    color: #f8fafc !important;
    font-size: 0.93rem !important;
    line-height: 1.75 !important;
}

textarea:focus, input:focus, select:focus {
    border-color: #38bdf8 !important;
    box-shadow: 0 0 0 4px rgba(56, 189, 248, 0.16) !important;
}

#gen-btn, #sum-btn, #submit-btn, #next-btn {
    border: none !important;
    border-radius: 16px !important;
    color: #ffffff !important;
    font-size: 0.96rem !important;
    font-weight: 800 !important;
    padding: 14px !important;
    width: 100% !important;
    box-shadow: 0 14px 35px rgba(56, 189, 248, 0.22) !important;
    transition: transform 0.15s ease, opacity 0.15s ease !important;
}

#gen-btn {
    background: linear-gradient(135deg, #38bdf8, #2563eb) !important;
}

#submit-btn {
    background: linear-gradient(135deg, #22c55e, #16a34a) !important;
}

#next-btn {
    background: linear-gradient(135deg, #8b5cf6, #6366f1) !important;
}

#sum-btn {
    background: linear-gradient(135deg, #f59e0b, #ef4444) !important;
}

#gen-btn:hover, #sum-btn:hover, #submit-btn:hover, #next-btn:hover {
    transform: translateY(-1px) !important;
    opacity: 0.92 !important;
}

.markdown, .prose {
    color: #e5e7eb !important;
}

.status-card {
    background: rgba(2, 6, 23, 0.55);
    border: 1px solid rgba(148, 163, 184, 0.20);
    border-radius: 18px;
    padding: 16px;
    color: #dbeafe;
    font-weight: 600;
}

.question-card {
    background: linear-gradient(135deg, rgba(30, 41, 59, 0.92), rgba(15, 23, 42, 0.92));
    border: 1px solid rgba(56, 189, 248, 0.22);
    border-radius: 20px;
    padding: 20px;
    color: #ffffff;
    font-size: 1.05rem;
    font-weight: 700;
    line-height: 1.8;
    box-shadow: inset 0 0 40px rgba(56, 189, 248, 0.04);
}

.feedback-card {
    background: rgba(15, 23, 42, 0.7);
    border: 1px solid rgba(148, 163, 184, 0.20);
    border-radius: 18px;
    padding: 14px;
    color: #fef3c7;
    font-weight: 700;
}

.footer-text {
    text-align: center;
    color: #94a3b8;
    font-size: 0.78rem;
    margin-top: 18px;
}

.tabs button {
    border-radius: 999px !important;
    font-weight: 800 !important;
}

.radio label {
    color: #e5e7eb !important;
    font-size: 0.95rem !important;
    text-transform: none !important;
    letter-spacing: 0 !important;
}

.container {
    border-radius: 18px !important;
}

::-webkit-scrollbar {
    width: 10px;
}

::-webkit-scrollbar-track {
    background: #020617;
}

::-webkit-scrollbar-thumb {
    background: #334155;
    border-radius: 999px;
}
"""

def is_arabic(text):
    count = sum(1 for c in text if '\u0600' <= c <= '\u06FF')
    return count > len(text) * 0.25


def extract_json_array(raw):
    start = raw.find("[")
    if start == -1:
        return None

    depth = 0
    for i in range(start, len(raw)):
        if raw[i] == "[":
            depth += 1
        elif raw[i] == "]":
            depth -= 1
            if depth == 0:
                return raw[start:i+1]

    return None


def validate_questions(questions):
    clean_questions = []

    if not isinstance(questions, list):
        return clean_questions

    for q in questions:
        if not isinstance(q, dict):
            continue

        if not all(k in q for k in ["question", "options", "correct"]):
            continue

        if not isinstance(q["options"], list) or len(q["options"]) < 2:
            continue

        try:
            q["correct"] = int(q["correct"])
        except:
            continue

        if q["correct"] < 0 or q["correct"] >= len(q["options"]):
            continue

        clean_questions.append(q)

    return clean_questions


def generate_quiz(text, question_type, num_questions):
    if not text.strip():
        return (
            [],
            0,
            0,
            "<div class='status-card'>⚠️ الرجاء إدخال نص أولاً.</div>",
            "<div class='question-card'>السؤال سيظهر هنا.</div>",
            gr.update(choices=[], value=None),
            ""
        )

    arabic = is_arabic(text)
    lang_note = "اكتب الأسئلة والخيارات باللغة العربية." if arabic else "Write questions and options in English."

    if question_type == "True/False":
        if arabic:
            fmt = '[{"question":"...","options":["صحيح","خطأ"],"correct":0}]'
        else:
            fmt = '[{"question":"...","options":["True","False"],"correct":0}]'
    else:
        fmt = '[{"question":"...","options":["A","B","C","D"],"correct":0}]'

    prompt = f"""أنت مولّد اختبارات محترف. {lang_note}
بناءً على النص التالي، أنشئ {int(num_questions)} سؤال من نوع {question_type}.

أعد JSON Array فقط.
لا تكتب أي شرح.
لا تكتب Markdown.
لا تكرر JSON أكثر من مرة.

الشكل المطلوب:
{fmt}

مهم:
- correct يجب أن يكون رقم الخيار الصحيح ويبدأ من 0.
- لا تكتب أي شيء قبل أو بعد JSON.
- الأسئلة يجب أن تكون واضحة ومباشرة.
- الخيارات يجب أن تكون منطقية.

النص:
{text}"""

    try:
        response = client.chat.completions.create(
            model="llama-3.1-8b-instant",
            messages=[
                {
                    "role": "system",
                    "content": "You only return valid JSON arrays. No explanations. No markdown."
                },
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            max_tokens=2000,
            temperature=0.2
        )

        raw = response.choices[0].message.content
        json_text = extract_json_array(raw)

        if not json_text:
            return (
                [],
                0,
                0,
                "<div class='status-card'>❌ خطأ في تحليل الإجابة، حاول مرة أخرى.</div>",
                "<div class='question-card'>السؤال سيظهر هنا.</div>",
                gr.update(choices=[], value=None),
                ""
            )

        questions = json.loads(json_text)
        questions = validate_questions(questions)

        if not questions:
            return (
                [],
                0,
                0,
                "<div class='status-card'>❌ لم يتم توليد أسئلة صالحة، حاول مرة أخرى.</div>",
                "<div class='question-card'>السؤال سيظهر هنا.</div>",
                gr.update(choices=[], value=None),
                ""
            )

        first_q = questions[0]
        question_text = f"<div class='question-card'>Q1. {first_q['question']}</div>"

        return (
            questions,
            0,
            0,
            f"<div class='status-card'>✅ تم توليد {len(questions)} سؤال بنجاح. ابدأ بالإجابة الآن.</div>",
            question_text,
            gr.update(choices=first_q["options"], value=None),
            "<div class='feedback-card'>اختر إجابة ثم اضغط Submit Answer.</div>"
        )

    except Exception as e:
        return (
            [],
            0,
            0,
            f"<div class='status-card'>❌ خطأ: {str(e)}</div>",
            "<div class='question-card'>السؤال سيظهر هنا.</div>",
            gr.update(choices=[], value=None),
            ""
        )


def submit_answer(answer, questions, index, score):
    if not questions:
        return "<div class='feedback-card'>⚠️ لا يوجد أسئلة بعد. اضغط Generate Quiz أولاً.</div>", score

    if answer is None:
        return "<div class='feedback-card'>⚠️ اختر إجابة أولاً.</div>", score

    if index >= len(questions):
        return "<div class='feedback-card'>🎉 انتهى الاختبار بالفعل.</div>", score

    current_q = questions[index]
    correct_index = current_q["correct"]
    correct_answer = current_q["options"][correct_index]

    if answer == correct_answer:
        score += 1
        return "<div class='feedback-card'>✅ إجابة صحيحة! ممتاز.</div>", score
    else:
        return f"<div class='feedback-card'>❌ إجابة خاطئة. الإجابة الصحيحة هي: {correct_answer}</div>", score


def next_question(questions, index, score):
    if not questions:
        return (
            index,
            "<div class='question-card'>السؤال سيظهر هنا.</div>",
            gr.update(choices=[], value=None),
            ""
        )

    index += 1

    if index >= len(questions):
        final_text = f"<div class='question-card'>🎉 انتهى الاختبار<br>نتيجتك النهائية: {score} / {len(questions)}</div>"
        return (
            index,
            final_text,
            gr.update(choices=[], value=None),
            "<div class='feedback-card'>عمل رائع! يمكنك توليد اختبار جديد من النص.</div>"
        )

    q = questions[index]
    question_text = f"<div class='question-card'>Q{index + 1}. {q['question']}</div>"

    return (
        index,
        question_text,
        gr.update(choices=q["options"], value=None),
        "<div class='feedback-card'>اختر إجابة ثم اضغط Submit Answer.</div>"
    )


def summarize_text(text):
    if not text.strip():
        return "⚠️ الرجاء إدخال نص أولاً."

    arabic = is_arabic(text)
    direction_note = "أجب باللغة العربية واستخدم النقاط المرقّمة." if arabic else "Answer in English using bullet points."

    prompt = f"""أنت مساعد ذكي متخصص في تلخيص النصوص. {direction_note}
قدّم:
1. ملخص موجز من 3 إلى 4 جمل
2. أهم النقاط على شكل قائمة
3. الكلمات المفتاحية

النص:
{text}"""

    try:
        response = client.chat.completions.create(
            model="llama-3.1-8b-instant",
            messages=[
                {
                    "role": "system",
                    "content": "You are a helpful text summarizer."
                },
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            max_tokens=1000,
            temperature=0.3
        )

        return response.choices[0].message.content

    except Exception as e:
        return f"❌ خطأ: {str(e)}"


with gr.Blocks(css=custom_css, title="AI Quiz Generator") as app:

    questions_state = gr.State([])
    index_state = gr.State(0)
    score_state = gr.State(0)

    gr.HTML("""
    <div id="hero">
        <div id="hero-content">
            <div class="logo-box">Q</div>
            <div>
                <h1 class="hero-title">AI Quiz Generator</h1>
                <p class="hero-subtitle">
                    Turn any Arabic or English study text into a smart interactive quiz,
                    then summarize it instantly using LLaMA 3.1 via Groq.
                </p>
                <div class="badges">
                    <span class="badge">Arabic + English</span>
                    <span class="badge">Interactive Quiz</span>
                    <span class="badge">AI Summary</span>
                    <span class="badge">Powered by Groq</span>
                </div>
            </div>
        </div>
    </div>
    """)

    with gr.Tabs():

        with gr.TabItem("🎯 Smart Quiz"):
            with gr.Row():
                with gr.Column(scale=5, elem_classes="panel"):
                    gr.Markdown("### 📘 Study Material")
                    text_input = gr.Textbox(
                        label="Input Text",
                        placeholder="Paste your lesson, notes, or paragraph here...",
                        lines=12
                    )

                    with gr.Row():
                        q_type = gr.Dropdown(
                            choices=["Multiple Choice (MCQ)", "True/False"],
                            value="Multiple Choice (MCQ)",
                            label="Question Type"
                        )

                        q_num = gr.Slider(
                            minimum=1,
                            maximum=10,
                            value=5,
                            step=1,
                            label="Number of Questions"
                        )

                    gen_btn = gr.Button("🚀 Generate Smart Quiz", elem_id="gen-btn")

                with gr.Column(scale=5, elem_classes="panel"):
                    gr.Markdown("### 🧠 Quiz Preview")
                    status = gr.HTML("<div class='status-card'>اضغط Generate Smart Quiz لتوليد الأسئلة.</div>")
                    question_box = gr.HTML("<div class='question-card'>السؤال سيظهر هنا.</div>")

                    answer_radio = gr.Radio(
                        choices=[],
                        label="Choose your answer",
                        interactive=True
                    )

                    with gr.Row():
                        submit_btn = gr.Button("✅ Submit Answer", elem_id="submit-btn")
                        next_btn = gr.Button("➡️ Next Question", elem_id="next-btn")

                    feedback = gr.HTML("<div class='feedback-card'>بانتظار توليد الأسئلة.</div>")

        with gr.TabItem("📄 AI Summarizer"):
            with gr.Row():
                with gr.Column(scale=5, elem_classes="panel"):
                    gr.Markdown("### ✍️ Text Input")
                    sum_input = gr.Textbox(
                        label="Text to Summarize",
                        placeholder="Paste your report, paragraph, or lesson here...",
                        lines=14
                    )

                    sum_btn = gr.Button("📋 Generate Summary", elem_id="sum-btn")

                with gr.Column(scale=5, elem_classes="panel"):
                    gr.Markdown("### 📝 Smart Summary")
                    sum_output = gr.Textbox(
                        label="Summary Output",
                        lines=17,
                        show_copy_button=True
                    )

    gr.HTML('<div class="footer-text">COMP 491 · Almaarefa University · AI Quiz Generator · LLaMA 3.1 via Groq</div>')

    gen_btn.click(
        fn=generate_quiz,
        inputs=[text_input, q_type, q_num],
        outputs=[
            questions_state,
            index_state,
            score_state,
            status,
            question_box,
            answer_radio,
            feedback
        ]
    )

    submit_btn.click(
        fn=submit_answer,
        inputs=[answer_radio, questions_state, index_state, score_state],
        outputs=[feedback, score_state]
    )

    next_btn.click(
        fn=next_question,
        inputs=[questions_state, index_state, score_state],
        outputs=[index_state, question_box, answer_radio, feedback]
    )

    sum_btn.click(
        fn=summarize_text,
        inputs=[sum_input],
        outputs=[sum_output]
    )

app.launch(share=True)
```
